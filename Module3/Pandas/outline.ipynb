# REGEX FILTERING

```python
import re
```

```python
data.loc[data['Name'].str.contains('Mega')].head(3)
```

```python
data.loc[data['Type 1'].str.contains('fire|grass', regex=True)].head(5)
```

```python
# Tắt chế độ case sensitive (viết hoa viết thường)
data.loc[data['Type 1'].str.contains('fire|grass', flags=re.I, regex=True)].head(5)
```

```python
# Tên chứa pi + 1 ký tự bất kỳ từ a - z + (0 hoặc nhiều ký tự bất kỳ)
data.loc[data['Name'].str.contains('pi[a-z]*', flags=re.I, regex=True)].head(5)
```

```python
# Bắt đầu bằng pi + 1 ký tự bất kỳ từ a - z + (0 hoặc nhiều ký tự bất kỳ)
data.loc[data['Name'].str.contains('^pi[a-z]*', flags=re.I, regex=True)].head(5)
```

```python
# Bắt đầu bằng pi + 1 ký tự bất kỳ (.) + (0 hoặc nhiều ký tự bất kỳ)
data.loc[data['Name'].str.contains('^pi.*', flags=re.I, regex=True)].head(5)
```

# BOX PLOT
1. Cấu tạo của Box Plot (5 thành phần chính)
- Hãy tưởng tượng bạn xếp tất cả dữ liệu từ nhỏ đến lớn.
    - Trung vị (Median / Q2): Đây là vạch kẻ nằm bên trong cái hộp.
        - Nó chia dữ liệu thành 2 phần bằng nhau (50% số liệu lớn hơn nó, 50% nhỏ hơn nó).
        - Tại sao quan trọng: Nó cho biết "giá trị trung tâm" thực sự, tốt hơn giá trị trung bình (mean) vì không bị ảnh hưởng bởi các giá trị quá lớn hoặc quá nhỏ.
    - Tứ phân vị thứ nhất (Q1 - 25%): Cạnh dưới (hoặc cạnh trái) của cái hộp.
        - 25% dữ liệu nhỏ nhất nằm dưới mức này.
    - Tứ phân vị thứ ba (Q3 - 75%): Cạnh trên (hoặc cạnh phải) của cái hộp.
        - 75% dữ liệu nằm dưới mức này (hoặc nói cách khác, 25% dữ liệu lớn nhất nằm trên mức này).
    - Hộp (The Box - IQR): Phần hình chữ nhật ở giữa.
        - Độ cao của hộp được gọi là IQR (Interquartile Range = Q3 - Q1).
        - Ý nghĩa: Cái hộp này chứa 50% lượng dữ liệu nằm ở giữa. Hộp càng ngắn thì dữ liệu càng tập trung, hộp càng dài thì dữ liệu càng phân tán.
