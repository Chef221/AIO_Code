# (1) BASIC PROBABILITY

**1. Basic Concepts**

- Experiment: implementation of set of basic conditions for observing a certain phenomenon
- An outcome is a result of an experiment
- The set of all possible outcomes is called the sample space
    + Roll a dice: Sample space $\Omega$ = {1, 2, 3, 4, 5, 6}
    + Toss a coin: Sample space $\Omega$ = {head, tail}
- An event is a subset of the sample space
- Random event: An event that may or may not occur when performing the experiment

- Example: A family with 2 children. Events:
    - A = 'A family has 1 boy and 1 girl': Random event
    - B = 'A family has 3 children': impossible event
    - C = 'A family has 2 children': Certain event

- Intersection of events:
    - In the experiment of rolling a single dice
        - Event A: 'The number rolled is even'
        $\rightarrow$ A = {2, 4, 6}
        - Event B: 'the number rolled is divisible by 3'
        $\rightarrow$ B = {3, 6}
        - A $\cap$ B = {6}

- Mutually exclusive event
    - Events A and B are mutually exclusive (cannot both occur at once) if they have no elements in common
    - For A and B have no outcomes in common means precisely that it is impossible for both A and B to occur on a single trial of the random experiment
    - A $\cap$ B = $\emptyset$

- Union of events
    - The collection of all outcomes that are elements of one or the other of the sets A and B, or of both of them
    - A $\cup$ B 

- Complements:
    - the complement of an event A in a sample space S, denoted A', or $A^c$
    - The collection of all outcomes in S that are not elements of the set A
    - A' + A = $\Omega$

- Definition
    - The probability of an event A is P(A)
    - P(A) => 0: very unlikely that the event A occurs
    - P(A) => 1: very likely to occur
    - Some properties:
        - 0 $\le$ P(A) $\le$ 1
        - P($\Omega$) = 1
        - P($\emptyset$) = 0

- Classical Probability
    - The theoretical probability of an event A is the number of ways the event can occur divided by the total number of possible outcomes
    $$P(A) = \frac{\text{number of favorable outcomes}}{\text{total number of possible outcomes}} = \frac{n_A}{n_\Omega}$$

- Example: Drawing a card from a well-shuffled deck. Find the probability of some events
    - A: 'Drawing a king from a deck of cards'
        - There are 52 cards in a deck of cards: $\rightarrow$ $n_\Omega$ = 52
        - There are 4 kings in a deck: $\rightarrow$ $n_A$ = 4
        - P(A) = $\frac{4}{52} = \frac{1}{13}$

    - B: 'Drawing a black card from a deck of cards'
        - There are 26 black cards: $\rightarrow$ P(B) = $\frac{26}{52}$ = 0.5

- Geometric Probability
    - When a variable is continuos, classical probability becomes impossible to 'count' the outcomes
    $$P(A) = \frac{\text{measure of domain A}}{\text{measure of domain }\Omega}$$

    - Example: X is a random real number between 0 and 3
    $$P(A) = \frac{\text{length of segment where 0 < X < 0.5}}{\text{length of segment where 0 < X < 3}} = \frac{0.5}{3} = \frac{1}{6}$$
    
    - Example: A dart is thrown at a circular dartboard such that it will land randomly over the area of the dartboard. What is the probability that it lands closer to the center 'success' than to the edge
        - A: 'A dart is thrown at the center area' $\rightarrow$ Measure of the areas in this 2D case:
        $$P(A) = \frac{\text{area of desired outcomes}}{\text{area of total outcomes}} = \frac{\frac{\pi r^2}{4}}{\pi r^2} = \frac{1}{4}$$

- The additive rule:
    - Mutually exclusive events:
    $$P(A+B) = P(A) + P(B)$$
    $$P(A \cup B) = P(A) + P(B)$$

    - In general
    $$P(A+B) = P(A) + P(B) - P(AB)$$
    $$P(A \cup B) = P(A) + P(B) - P(A \cap B)$$

- Probability for complements
    - For any event A
    $$P(A^c) = 1 - P(A)$$
    $$P(A) = 1 - P(A^c)$$

- Conditional Probability
    - Formula:
    $$P(A | B) = \frac{P(A \cap B)}{P(B)}$$
    $$P(A \cap B) = P(B)P(A | B)$$

    - Example: A fair dice is rolled
    - Find the probability that the number rolled is a five, given that it is odd
        - Sample space S = {1, 2, 3, 4, 5, 6}, consisting of 6 equally likely outcomes
        - A: 'a five is rolled' $\rightarrow$ P(A) = 1/6
        - B: 'an odd number is rolled' $\rightarrow$ P(B) = 3/6 = 1/2
        - A and B = {5} $\rightarrow$ P(A \cap B) = 1/6
        $$\rightarrow P(A|B) = \frac{P(A \cap B)}{P(B)} = \frac{1/6}{1/2} = 1/3$$

    - Find the probability that the number rolled is odd, given that it is a five
        - P(A) = 1/6
        - P(B) = 3/6 = 1/2
        - P(A $\cap$ B) = 1/6
        $$\rightarrow P(B|A) = \frac{P(A \cap B)}{P(A)} = \frac{1/6}{1/6} = 1$$

- Multiplication rule:
$$P(AB) = P(A)P(B|A) = P(B)P(A|B)$$
$$P(A_1A_2...A_n) = P(A_1)P(A_2|A_1)P(A_3|A_1A_2)...P(A_n|A_1A_2...A_{n-1})$$

- Independent events:
    - Events A and B are independent if:
    $$P(AB) = P(A)P(B)$$

- Total Probability Theorem
    - Let $A_1, A_2,...,A_n$ - complete system of events. Consider any event H such that H occurs only when one of the events $A_1, A_2,...,A_n$ occurred
    $$P(H) = P(H_1) + P(H_2) + P(H_3)$$
    $$P(H) = P(A_1)P(H|A_1) + P(A_2)P(H|A_2) + P(A_3)P(H|A_3)$$

**2. Bayes' Theorem**

- For any two events A and B, where P(A) $\ne$ 0:
$$P(B|A) = \frac{P(A|B)P(B)}{P(A)}$$
- P(A|B): Likelihood (The probability of 'A' being True, given 'B' True)
- P(B): Prior (The probability of 'B' being True. This is the knowledge)
- P(B|A): Posterior (The probability of 'B' being True. Given 'A' True)
- P(A): Marginalization (The probability of 'A' being True)

- If $C_1, C_2,...,C_n$: complete system of events and X is any event with P(X) $\ne$ 0
$$P(C_i|X) = \frac{P(C_i)P(X|C_i)}{P(X)} = \frac{P(C_i)P(X|C_i)}{\sum_{j = 1}^nP(C_j)P(X|C_j)}$$

- Example: Detect Spam E-Mail
    - Assume that the word 'offer' occurs in 80% of the spam messages in my account. Also, let's assume 'offer' occurs in 10% of my desired emails. If 30% of the received emails are considered as a spam. I will receive a new message which contains 'offer', what is the probability that it is spam ?

    - A: 'email contains 'offer''
    - B: 'email is spam'
    - We need to calculate: P(B|A)

        - P(A|B): 0.8
        - P(B) = 0.3 => P(Not B) = 1 - 0.3 = 0.7
        - P(A) = P(B)P(A|B) + P(Not B)P(A|Not B) = 0.3 x 0.8 + 0.7 x 0.1 = 0.31
    $$\rightarrow P(B|A) = \frac{P(A|B)P(B)}{P(A)} = \frac{0.8 * 0.3}{0.31} = 0.774$$
