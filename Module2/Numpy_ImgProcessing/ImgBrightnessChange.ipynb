Do quy ước ngược nhau giữa hàm vẽ (OpenCV) và cách lưu trữ dữ liệu (NumPy).
- cv2.resize: Nhận tham số theo hệ toạ độ hình học (Rộng, Cao) hay $(x, y)$.
- img.shape: Trả về tham số theo hệ ma trận (Hàng, Cột) hay $(Height, Width)$.

1. Hàm cv2.resize (Tư duy Hình học)

- Khi bạn dùng các hàm chỉnh sửa hình học trong OpenCV (như resize, line, circle...), thư viện này tư duy theo hệ toạ độ Descartes $(x, y)$ giống như màn hình máy tính:
    - Tham số thứ nhất: Width (Chiều rộng - trục X).
    - Tham số thứ hai: Height (Chiều cao - trục Y).
=> cv2.resize(img, (600, 400)) nghĩa là bạn muốn ảnh mới có:
    - Chiều rộng (Width) = 600
    - Chiều cao (Height) = 400

2. Thuộc tính .shape (Tư duy Ma trận)

- Ảnh trong Python được lưu dưới dạng mảng NumPy (Matrix). Trong đại số tuyến tính, ma trận luôn được gọi theo thứ tự Hàng trước, Cột sau ($Row \times Column$).
    - Hàng (Row): Tương ứng với chiều cao (Height).
    - Cột (Column): Tương ứng với chiều rộng (Width).
=> img.shape sẽ trả về (Height, Width, Channels). Vì bạn đã set Height = 400 ở trên, nên nó sẽ hiện số 400 trước.

```python
import cv2
import matplotlib.pyplot as plt

img1 = cv2.resize(img1, (600, 400))
print(img1.shape)
# Giả sử ảnh có kích thước (400, 600, 3)
# Trong OpenCV thứ tự là BGR (Blue - Green - Red)

# 1. Bóc tách từng lớp (Tách thành 3 ma trận 400x600 riêng biệt)
blue_channel, green_channel, red_channel = cv2.split(img1)

# Kiểm tra kích thước
print(blue_channel.shape)  # Kết quả sẽ là (400, 600) -> Đây chính là 1 ma trận 2D
print(green_channel.shape) # (400, 600)
print(red_channel.shape)   # (400, 600)

# 2. Hiển thị thử kênh màu đỏ (sẽ hiển thị dưới dạng ảnh đen trắng)
plt.imshow(red_channel, cmap='gray')
plt.title("Kênh màu Đỏ (Red Channel)")
plt.show()
```

Đây là hai cách nhìn khác nhau về cùng một cục dữ liệu (khối 3 chiều), tuỳ thuộc vào việc bạn đang đứng ở góc độ nào để "cắt" lớp dữ liệu đó.

1. Góc nhìn Lập trình/Cấu trúc mảng

- Trong Python (NumPy), mảng nhiều chiều thực chất là các mảng lồng nhau (List of Lists of Lists). Hãy nhìn vào cấu trúc (400, 600, 3):
    - Lớp ngoài cùng (400): Tượng trưng cho 400 dòng (chiều cao).
    - Mỗi phần tử trong 400 dòng đó: Là một ma trận con kích thước 600 x 3.
- Hãy tưởng tượng bức ảnh là một cuốn sách:
    - Cuốn sách có 400 trang (tương ứng 400 dòng ảnh).
    - Mỗi trang giấy chứa một bảng dữ liệu kích thước 600 hàng x 3 cột.
        - 600 ở đây là chiều ngang (số điểm ảnh trên 1 dòng).
        - 3 ở đây là màu sắc (Blue, Green, Red) của từng điểm đó.

```python
import numpy as np

# Tạo một ảnh giả lập kích thước 400x600x3
img = np.zeros((400, 600, 3))

# Lấy ra phần tử đầu tiên (dòng số 0)
dong_dau_tien = img[0]

print(f"Kích thước dòng đầu tiên: {dong_dau_tien.shape}")
# Kết quả sẽ in ra: (600, 3)
```

2. Sự khác biệt giữa 2 cách hiểu

- Cách hiểu 1: "3 ma trận 400 x 600 xếp chồng lên nhau"
    - Góc nhìn: Xử lý màu sắc (Color Channels).
    - Hình dung: Giống như chiếc bánh kẹp 3 lớp (Lớp thịt, lớp rau, lớp vỏ).
    - Ứng dụng: Khi bạn muốn tách riêng màu đỏ ra để xử lý, hoặc làm xám ảnh.
    - Lệnh: cv2.split(img).
- Cách hiểu 2: "400 dải băng, mỗi dải kích thước 600 x 3" (Cách bạn vừa hỏi)
    - Góc nhìn: Cấu trúc dữ liệu máy tính (Memory Layout/Indexing).
    - Hình dung: Giống như cái mành tre. Cả cái mành có 400 thanh tre nằm ngang. Mỗi thanh tre là một dãy các điểm màu dài (600, 3).
    - Ứng dụng: Khi máy tính đọc ảnh, nó đọc từng dòng một từ trên xuống dưới (quét ngang).
    - Lệnh: img[i] (truy cập dòng thứ i).

```python
import cv2
import numpy as np

# 1. Tạo một ảnh đen hoàn toàn, kích thước 400x600x3
# dtype=np.uint8 là kiểu dữ liệu chuẩn cho ảnh (0-255)
img = np.zeros((400, 600, 3), dtype=np.uint8)

print(f"Kích thước ảnh ban đầu: {img.shape}")

# --- PHẦN QUAN TRỌNG NHẤT ---

# Truy cập vào dòng thứ 200 (index là 199)
# Theo cách hiểu này, img[199] chính là một ma trận con kích thước (600, 3)
dong_thu_200 = img[199]
print(f"Kích thước của riêng dòng thứ 200: {dong_thu_200.shape}")

# Gán màu ĐỎ cho TOÀN BỘ dòng này.
# OpenCV dùng hệ màu BGR -> Màu đỏ là [0, 0, 255] (Xanh dương=0, Xanh lá=0, Đỏ=255)
# Nhờ cơ chế thông minh của NumPy (broadcasting), ta không cần vòng lặp for.
img[199] = [0, 0, 255]

# Cách CHẬM (tư duy từng điểm ảnh)
for cot in range(600):
    img[199, cot] = [0, 0, 255]

# 3. Hiển thị kết quả
if img is not None:
    # 2. Chuyển đổi màu từ BGR sang RGB
    image_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    # 3. Hiển thị bằng Matplotlib
    plt.imshow(image_rgb)
    plt.axis('off') 
    plt.show()
else:
    print("Không tìm thấy ảnh")
```

# Application: Increase the brightness of a grayscale image

- Idea: For each pixel, increase pixel value by a value v

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread(r'C:\Code_AI\AIO_Code\Module2\Numpy_ImgProcessing\cat1.jpg')

if img is not None:
    # 2. Chuyển đổi màu từ BGR sang RGB
    image_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    # 3. Hiển thị bằng Matplotlib
    plt.imshow(image_rgb)
    plt.axis('off') 
    plt.show()
else:
    print("Không tìm thấy ảnh")
```

```python
h, w, c = img.shape
new_img = np.zeros((h, w, c), dtype=np.uint8)

for i in range(h):
    for j in range(w):
        for k in range(c):
            gia_tri_moi = int(img[i, j, k]) + 100
            
            if gia_tri_moi > 255:
                gia_tri_moi = 255
            elif gia_tri_moi < 0:
                gia_tri_moi = 0

            # 3. Gán lại vào ảnh mới
            new_img[i, j, k] = gia_tri_moi

if new_img is not None:
    # 2. Chuyển đổi màu từ BGR sang RGB
    image_rgb = cv2.cvtColor(new_img, cv2.COLOR_BGR2RGB)

    # 3. Hiển thị bằng Matplotlib
    plt.imshow(image_rgb)
    plt.axis('off') 
    plt.show()
else:
    print("Không tìm thấy ảnh")
```

```python
h, w, c = img.shape
new_img_2 = np.zeros((h, w, c), dtype=np.uint8)

for i in range(h):
    for j in range(w):
        for k in range(c):
            new_val = int(img[i, j, k]) - 100
            if new_val < 0:
                new_val = 0
            elif new_val > 255:
                new_val = 255
            new_img_2[i, j, k] = new_val

if new_img_2 is not None:
    # 2. Chuyển đổi màu từ BGR sang RGB
    image_rgb = cv2.cvtColor(new_img_2, cv2.COLOR_BGR2RGB)

    # 3. Hiển thị bằng Matplotlib
    plt.imshow(image_rgb)
    plt.axis('off') 
    plt.show()
else:
    print("Không tìm thấy ảnh")
```

# Application: Background Subtraction

```python
import cv2 
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread(r'C:\Code_AI\AIO_Code\Module2\Numpy_ImgProcessing\images.jpg')
if img is not None:
    # 2. Chuyển đổi màu từ BGR sang RGB
    image_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    # 3. Hiển thị bằng Matplotlib
    plt.imshow(image_rgb)
    plt.axis('off') 
    plt.show()
else:
    print("Không tìm thấy ảnh")
```

```python
h, w, c = img.shape
green_background = np.zeros((h, w, c), dtype=np.uint8)

green_background[:] = (0, 255, 0)
cv2.imshow('Green Background', green_background)

cv2.imwrite('green_bg.jpg', green_background) 
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
subtract_cat = np.zeros((h,w,c), dtype=np.uint8)

subtract_cat = cv2.subtract(img, green_background)
if subtract_cat is not None:
    # 2. Chuyển đổi màu từ BGR sang RGB
    image_rgb = cv2.cvtColor(subtract_cat, cv2.COLOR_BGR2RGB)

    # 3. Hiển thị bằng Matplotlib
    plt.imshow(image_rgb)
    plt.axis('off') 
    plt.show()
else:
    print("Không tìm thấy ảnh")
```

```python
print(subtract_cat.shape)
sky_img = cv2.imread(r'C:\Code_AI\AIO_Code\Module2\Numpy_ImgProcessing\images.png')
sky_img = cv2.resize(sky_img, (194, 259))
print(sky_img.shape)
if sky_img is not None:
    # 2. Chuyển đổi màu từ BGR sang RGB
    image_rgb = cv2.cvtColor(sky_img, cv2.COLOR_BGR2RGB)

    # 3. Hiển thị bằng Matplotlib
    plt.imshow(image_rgb)
    plt.axis('off') 
    plt.show()
else:
    print("Không tìm thấy ảnh")
```

```python
new_img = cv2.add(subtract_cat, sky_img)
if new_img is not None:
    # 2. Chuyển đổi màu từ BGR sang RGB
    image_rgb = cv2.cvtColor(new_img, cv2.COLOR_BGR2RGB)

    # 3. Hiển thị bằng Matplotlib
    plt.imshow(image_rgb)
    plt.axis('off') 
    plt.show()
else:
    print("Không tìm thấy ảnh")
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

color_img = cv2.imread(r'C:\Code_AI\AIO_Code\Module2\Numpy_ImgProcessing\images.jpg')
if color_img is not None:
    # 2. Chuyển đổi màu từ BGR sang RGB
    image_rgb = cv2.cvtColor(color_img, cv2.COLOR_BGR2RGB)

    # 3. Hiển thị bằng Matplotlib
    plt.imshow(image_rgb)
    plt.axis('off') 
    plt.show()
else:
    print("Không tìm thấy ảnh")
```

```python
# Vector chuyển đổi [B, G, R]
vector = np.array([0.072, 0.715, 0.213])

# 1. Nhân ma trận để tạo ảnh xám
# Kết quả trả về là mảng float64
grayscale_img = color_img @ vector

# 2. Ép kiểu về uint8
grayscale_img = grayscale_img.astype(np.uint8)

# 3. Hiển thị
plt.imshow(grayscale_img, cmap='gray')
plt.axis('off')
plt.show()

print("Shape của ảnh sau khi nhân:", grayscale_img.shape) 
# Kết quả sẽ là (H, W) -> 2 chiều, không còn số 3 ở cuối
```
