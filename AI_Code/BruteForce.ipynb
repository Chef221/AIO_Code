# (1) Brute Force
**1.1. Brute Force approach**
- A straightforward approach to solving a given problem (solving a given problem in the most simple, or obvious)
- Directly based on the problem statement and definitions of the concepts involved
- Not considered efficient method of solving a problem

**1.2. Sorting problem**
- Straightforward way of solving sorting problem
    - Moving the smaller elements to the first positional in the sequence (Selection Sort)
    - Moving the larger elements to the last positional in the sequence (Bubble Sort)



```python
def selectionSort(arr):
    n = len(arr)
    for i in range(n):
        min_idx = i
        for j in range(i + 1, n):
            if arr[j] < arr[min_idx]:
                min_idx = j
        if (min_idx != i):
            arr[i], arr[min_idx] = arr[min_idx], arr[i]
    return arr

def bubbleSort(arr):
    n = len(arr)
    for i in range(1, n):
        for j in range(0, n - i - 1):
            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
    return arr

a = [1, 3, 2, 4, 5, 3]
print(bubbleSort(a))
print(selectionSort(a))

```

# (2) Exhaustive Search
- A brute force solution to a problem involving search for an element with a special property or constraint by exploring every possible combination from a set of choices or values
- Method:
    - Create all possible solutions in a systematic manner
    - Evaluate potential solutions one by one, remove infeasible ones and keep track of the best one found so far
    - When search ends, output the solution found

**1. Travelling Salesman Problem**
- Bước 1 - Sinh tất cả lộ trình (Hoán vị): Giả sử có 4 thành phố A, B, C, D xuất phát từ A. Máy tính sẽ liệt kê tất cả các thứ tự đi có thể:
    - A -> B -> C -> D -> A
    - A -> B -> D -> C -> A
    - ... và cứ thế liệt kê hết $(n-1)!$ trường hợp.
- Bước 2 - Tính toán và So sánh:
    - Tính tổng độ dài quãng đường cho lộ trình 1. Gán nó là min_distance.
    - Tính tổng độ dài lộ trình 2. Nếu nó ngắn hơn min_distance, cập nhật min_distance mới.
    - Lặp lại cho đến hết các lộ trình.
- Đánh giá độ phức tạp:
Vì phải xét tất cả các hoán vị của $n$ thành phố, độ phức tạp của thuật toán này là $O(n!)$ (Giai thừa).
    - Với $n=10$, máy tính chạy rất nhanh.
    - Với $n=50$, số trường hợp lớn hơn số nguyên tử trong vũ trụ, máy tính không thể chạy xong trong hàng triệu năm.

$\rightarrow$ Đây là ví dụ điển hình cho thấy Exhaustive Search chính xác tuyệt đối nhưng chi phí cực cao.

```python
from sys import maxsize
from itertools import permutations

def travelling_salesman_problem(distances, s, num_cities):
    cities = []
    for i in range(num_cities):
        if i != s:
            cities.append(i)
    min_path = maxsize
    next_permutation = permutations(cities)
    for i in next_permutation:
        current_pathweight = 0
        k = s
        for j in i:
            current_pathweight += distances[k][j]
            k = j
        current_pathweight += distances[k][s]
        min_path = min(min_path, current_pathweight)
    return min_path
```
