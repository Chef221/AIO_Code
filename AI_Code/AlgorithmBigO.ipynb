# The seven most important functions
(1) The constant function: $f(n) = c$
    - Any argument n, $f(n)$ assigns the value $c$
    - Basic operation: comparing two numbers,...

(2) The logarithm function: $f(n)$ = $\log_b(n)$ (in computer science, b = 2)

(3) The linear function: $f(n) = n$
    - The best running time to achieve for any algorithm
    - Comparing a number x to each element of a sequence of size n

(4) The N-Log-N function: $f(n) = n\log(n)$
    - The fastest possible algorithms for sorting

(5) The Quadratic function: $f(n) = n^2$
    - Appears often in algorithm analysis: nested loops

(6) The cubic function: $f(n) = n^3$

(7) The exponential function: $f(n) = b^n$

# (3) "Big-O" Notation
- $f(n)$ là $O(g(n))$ nếu tồn tại hằng số $c > 0$ và $n_0$ sao cho:
$f(n) \leq c \cdot g(n)$ với mọi $n \geq n_0$
    - $f(n)$: Là chi phí thực tế của thuật toán (ví dụ: $8n + 5$).
    - $g(n)$: Là hàm đơn giản hóa đại diện cho độ phức tạp (ví dụ: $n, n^2$).
    - $c \cdot g(n)$: Là cái trần nhà.
    - $n_0$: Là thời điểm bắt đầu tính. Big-O chỉ quan tâm đến khi n rất lớn (về lâu dài). Có thể lúc đầu ($n$ nhỏ) $f(n)$ trồi sụt lung tung, nhưng kể từ điểm $n_0$ trở đi, $f(n)$ luôn luôn nằm dưới cái trần $c \cdot g(n)$.



- Ý nghĩa thực tế: Bạn hãy tưởng tượng "Big-O" giống như việc tìm một "cái trần nhà" (giới hạn trên) cho hàm số của bạn.


Để chứng minh $f(n) = 8n + 5$ thuộc về lớp $O(n)$, ta cần áp dụng đúng công thức định nghĩa của Big-O.
- Ở đây, ta xác định các thành phần như sau:
    - Hàm thực tế: $f(n) = 8n + 5$
    - Hàm tiệm cận (hàm ta nghi ngờ): $g(n) = n$
- Nhiệm vụ của ta là đi tìm hai số hằng số $c$ và $n_0$ để thỏa mãn bất đẳng thức:
$$f(n) \leq c \cdot g(n)$$
$$\Leftrightarrow 8n + 5 \leq c \cdot n$$


- Bước 1: Quan sát và đánh giá
    - Ta thấy vế trái ($8n + 5$) có bậc cao nhất là 1 (tức là $n^1$). Vế phải ($c \cdot n$) cũng là bậc 1. Điều này cho thấy hướng đi $O(n)$ là đúng.
Mục tiêu bây giờ là biến đổi số $5$ (hằng số) thành một cái gì đó chứa $n$ để có thể cộng gộp với $8n$.
- Bước 2: Chọn $n_0$ (Điểm bắt đầu)
    - Ta cần chọn một điểm mốc $n_0$ để bắt đầu xét. Thông thường chọn $n_0 = 1$ cho dễ.
    - Nếu ta chọn $n_0 = 1$, nghĩa là ta đang xét các trường hợp $n \geq 1$.
Khi $n \geq 1$, điều hiển nhiên là:
    $$5 \leq 5 \cdot n$$

- Bước 3: Thay thế và tìm $c$
    - Từ bất đẳng thức trên, ta thay thế vào hàm gốc $f(n)$:
    $$f(n) = 8n + 5$$
    $$\leq 8n + 5n$$
    $$= 13n$$
    - Nhìn lại công thức định nghĩa: $f(n) \leq c \cdot n$.
Ở đây ta vừa chứng minh được: $f(n) \leq 13 \cdot n$.

# (4) "Big-Omega" Notation
- Nếu như trong phần trước, Big-O là "cận trên" (giới hạn trần - thể hiện trường hợp tệ nhất hoặc tối đa), thì Big-Omega là ngược lại, nó là "cận dưới" (giới hạn sàn).

**1. Định nghĩa toán học**
- Ý nghĩa: Hàm $f(n)$ thuộc $\Omega(g(n))$ nếu tồn tại hằng số $c > 0$ và $n_0$ sao cho:
$$f(n) \geq c \cdot g(n)$$
- Từ khóa quan trọng: "greater-than-or-equal-to" (Lớn hơn hoặc bằng).
- Thực tế: Điều này có nghĩa là thuật toán của bạn sẽ tốn thời gian ít nhất là chừng này, nó không thể chạy nhanh hơn mức giới hạn này. Đường $f(n)$ luôn nằm trên đường $c \cdot g(n)$ khi $n$ đủ lớn.

**2. Ví dụ chứng minh**
Bài toán yêu cầu chứng minh hàm $f(n) = 3n\log n - 2n$ thuộc lớp $\Omega(n\log n)$.
- Cách giải:
Người ta cần tìm $c$ và $n_0$ sao cho $3n\log n - 2n \geq c \cdot n\log n$.
- Biến đổi biểu thức:
Tách $3n\log n$ thành $1n\log n + 2n\log n$ để dễ so sánh.
    $$3n\log n - 2n = n\log n + 2n\log n - 2n$$
    $$= n\log n + 2n(\log n - 1)$$
- Đánh giá:
Ta muốn cả biểu thức trên lớn hơn hoặc bằng $n\log n$ (tức là chọn ngầm định $c=1$).
Để điều đó xảy ra, phần dư ra là $2n(\log n - 1)$ phải không được âm ($\geq 0$).
- Tìm điều kiện ($n_0$):
Để $2n(\log n - 1) \geq 0$, ta cần:
$$\log n - 1 \geq 0$$
$$\Leftrightarrow \log n \geq 1$$
$$\Leftrightarrow n \geq 2$$
- Kết luận:
Vì với $n \geq 2$, biểu thức luôn lớn hơn $1 \cdot n\log n$.
Nên ta chọn được: $c = 1$ và $n_0 = 2$.


**3. Tóm tắt sự khác biệt:**
- Big-O ($O$): $f(n) \leq c \cdot g(n)$ $\rightarrow$ Tìm chặn trên (Máy chạy không quá thời gian này).
- Big-Omega ($\Omega$): $f(n) \geq c \cdot g(n)$ $\rightarrow$ Tìm chặn dưới (Máy chạy ít nhất thời gian này).

# (5) "Big-Theta" Notation
Big-O (cận trên - trần nhà) và Big-Omega (cận dưới - sàn nhà), thì Big-Theta chính là sự kết hợp của cả hai. Đây là khái niệm quan trọng nhất để mô tả độ phức tạp "chính xác" của một thuật toán.

**1. Định nghĩa "Big-Theta" (Cận chặt)**
- Ký hiệu: $f(n) = \Theta(g(n))$
- Ý nghĩa toán học: Hàm $f(n)$ thuộc lớp $\Theta(g(n))$ nếu tồn tại 3 con số: hai hằng số dương $c_1, c_2$ và một điểm mốc $n_0$ sao cho:
$$c_1 \cdot g(n) \le f(n) \le c_2 \cdot g(n)$$
- Hiểu đơn giản (Nguyên lý Sandwich):
    - Hàm thực tế $f(n)$ bị kẹp giữa hai đường: đường sàn $c_1 \cdot g(n)$ và đường trần $c_2 \cdot g(n)$.
    - Điều này có nghĩa là $f(n)$ tăng trưởng cùng tốc độ với $g(n)$. Nó không chạy nhanh hơn, cũng không chạy chậm hơn.
    - Slide có ghi một câu chốt rất hay: $f(n)$ là $\Theta(g(n))$ NẾU $f(n)$ vừa là $O(g(n))$ VÀ vừa là $\Omega(g(n))$.

**2. Ví dụ chứng minh**
Đề bài: Chứng minh $f(n) = 3n\log n + 4n + 5\log n$ thuộc lớp $\Theta(n\log n)$.
Ta cần tìm $c_1, c_2$ và $n_0$.
- Cách giải (Phương pháp đánh giá): Ta cần chứng minh bất đẳng thức kép:
$$c_1(n\log n) \le \underbrace{3n\log n + 4n + 5\log n}_{f(n)} \le c_2(n\log n)$$
- Bước 1: Tìm $c_1$ (Đánh giá cận dưới - Tìm cái nhỏ hơn)
    - Ta thấy các số hạng $4n$ và $5\log n$ đều là số dương (với $n$ đủ lớn).
    - Nếu bỏ bớt số hạng dương đi thì biểu thức sẽ nhỏ lại.
    - Suy ra: $3n\log n \le 3n\log n + 4n + 5\log n$.
    - Vậy ta chọn được $c_1 = 3$.
- Bước 2: Tìm $c_2$ (Đánh giá cận trên - Tìm cái lớn hơn)
    - Ta thay thế các số hạng bậc thấp ($4n$ và $5\log n$) bằng số hạng bậc cao nhất ($n\log n$) để làm biểu thức lớn lên.
    - $4n \le 4n\log n$ (Đúng với $n \ge 2$).
    - $5\log n \le 5n\log n$ (Hiển nhiên).
    - Cộng lại: $3n\log n + 4n + 5\log n \le (3 + 4 + 5)n\log n = 12n\log n$.
    - Vậy ta chọn được $c_2 = 12$.
- Bước 3: Điều kiện $n_0$
    - Để các phép so sánh trên đúng (ví dụ $n \le n\log n$), ta cần $\log n \ge 1$, tức là $n \ge 2$ (nếu log cơ số 2).
    - Vậy $n_0 = 2$.

**3. Tóm lại**
- Khi nói một thuật toán có độ phức tạp $\Theta(n^2)$, nghĩa là nó chắc chắn chạy mất khoảng bình phương thời gian, bất kể dữ liệu đầu vào đẹp hay xấu, nó luôn dao động trong khoảng đó chứ không nhanh hơn hay chậm hơn đột biến. Đây là đánh giá chặt chẽ nhất.

**a) $(n + 1)^5$ is O($n^5$)**

- Để chứng minh điều này, theo định nghĩa Big-O, ta cần tìm hai hằng số dương $c$ và $n_0$ sao cho:
$$(n+1)^5 \le c \cdot n^5$$

- Chọn điểm mốc $n_0$:
Giả sử ta xét với mọi $n \ge 1$ (tức là chọn $n_0 = 1$).
- Đánh giá biểu thức $(n+1)$:
Vì $n \ge 1$, nên ta có bất đẳng thức: $1 \le n$.
Suy ra:
$$n + 1 \le n + n = 2n$$
- Mũ 5 lên cả hai vế:
Từ $n + 1 \le 2n$, ta mũ 5 cả hai vế (vì các số đều dương nên chiều bất đẳng thức giữ nguyên):
$$(n+1)^5 \le (2n)^5$$
- Tính toán:
$$(2n)^5 = 2^5 \cdot n^5 = 32 \cdot n^5$$
- Kết luận:Ta có: $(n+1)^5 \le 32 \cdot n^5$ (với mọi $n \ge 1$).So sánh với công thức định nghĩa $c \cdot n^5$, ta tìm được:
    - $c = 32$
    - $n_0 = 1$

**b) $2^{n + 1}$ is O($2^n$)**

Theo định nghĩa Big-O, ta cần tìm hằng số $c > 0$ và $n_0$ sao cho:
$$2^{n+1} \le c \cdot 2^n$$

- Bước 1: Biến đổi biểu thức vế trái
Ta sử dụng tính chất cơ bản của lũy thừa: $a^{m+n} = a^m \cdot a^n$.
Áp dụng vào bài toán:
$$2^{n+1} = 2^1 \cdot 2^n = 2 \cdot 2^n$$
- Bước 2: So sánh để tìm $c$
Ta thay kết quả vừa biến đổi vào bất đẳng thức định nghĩa Big-O:
$$2 \cdot 2^n \le c \cdot 2^n$$
Để bất đẳng thức này đúng, ta chỉ cần chọn $c = 2$.
Khi đó:
$$2 \cdot 2^n \le 2 \cdot 2^n$$
- Bước 3: Chọn $n_0$
Vì đẳng thức trên đúng với mọi $n$, nên ta có thể chọn điểm bắt đầu $n_0 = 1$ (hoặc bất kỳ số nguyên nào).

**c) $5n^2 + 3n\log n + 2n + 5$ là $O(n^2)$.**

- Bước 1: Chọn điểm mốc $n_0$
Chọn $n_0 = 1$ (tức là xét với mọi $n \ge 1$).
- Bước 2: Đánh giá từng thành phần
Với $n \ge 1$, ta luôn có các bất đẳng thức sau:
    - $5n^2 \le 5n^2$ (Giữ nguyên).
    - $3n\log n \le 3n^2$ (Giải thích: Vì $\log n < n$ với mọi $n \ge 1$, nên nhân cả hai vế với $3n$ ta được $3n\log n < 3n^2$).
    - $2n \le 2n^2$ (Giải thích: Vì $n \le n^2$ với mọi $n \ge 1$).
    - $5 \le 5n^2$ (Giải thích: Vì $1 \le n^2$ với mọi $n \ge 1$).
- Bước 3: Cộng gộp để tìm $c$
Thay thế các bất đẳng thức trên vào hàm số ban đầu:
$$\begin{aligned}
f(n) &= 5n^2 + 3n\log n + 2n + 5 \\
&\le 5n^2 + 3n^2 + 2n^2 + 5n^2 \\
&= (5 + 3 + 2 + 5)n^2 \\
&= 15n^2
\end{aligned}$$

So sánh với định nghĩa $c \cdot n^2$, ta chọn được $c = 15$.

**d) $3\log(n) + 2$ is O($\log(n)$)**

- Bước 1: Chọn điểm mốc $n_0$
Trong khoa học máy tính, $\log n$ thường được hiểu là logarit cơ số 2.
Ta chọn $n_0 = 2$.
Tại sao lại là 2? Vì khi $n \ge 2$, ta có $\log_2 n \ge 1$. Điều này giúp ta dễ dàng so sánh số hằng với $\log n$.
- Bước 2: Đánh giá (Làm trội)
Ta cần biến đổi số 2 thành một số hạng chứa $\log n$.
Với mọi $n \ge 2$, ta có:
$$1 \le \log n$$
Nhân cả hai vế với 2, ta được:
$$2 \le 2\log n$$
- Bước 3: Thay thế và cộng gộp Thay bất đẳng thức trên vào biểu thức ban đầu:
$$\begin{aligned}
f(n) &= 3\log n + 2 \\
&\le 3\log n + 2\log n \\
&= 5\log n
\end{aligned}$$
So sánh với định nghĩa $c \cdot \log n$, ta chọn được $c = 5$.

**f) $n^2$ is $\Omega(n\log(n))$**

Theo định nghĩa, ta cần tìm hằng số $c > 0$ và $n_0$ sao cho:
$$n^2 \ge c \cdot n\log n$$

- Bước 1: Phân tích và đơn giản hóa bất đẳng thức
Ta cần tìm $c$ để bất đẳng thức sau đúng:
$$n^2 \ge c \cdot n\log n$$
Vì $n$ là số dương (khi xét $n \ge 1$), ta có thể chia cả hai vế cho $n$ để đơn giản hóa:
$$n \ge c \cdot \log n$$
- Bước 2: Chọn hằng số $c$
Ta biết rằng hàm tuyến tính ($n$) luôn tăng trưởng nhanh hơn hàm logarit ($\log n$).
Ví dụ:
    - Với $n=2$: $2 > \log_2(2) = 1$

A. Tổng chi phí vòng lặp trong (Inner Loop):
Bao gồm: (Số lần chạy thân hàm $\times$ 4 op) + (Số lần kiểm tra sai $\times$ 1 op)
$$= \left[ 4 \times \frac{n(n+1)}{2} \right] + [ 1 \times n ]$$
B. Tổng chi phí vòng lặp ngoài (Outer Loop):
Bao gồm: (Số lần chạy $\times$ 2 op) + (1 lần kiểm tra sai cuối cùng)
$$= [ 2 \times n ] + 1$$
$\Rightarrow$ TỔNG CỘNG (A + B):
$$\text{Total} = \left( 2n^2 + 2n \right) + n + 2n + 1$$
$$\text{Total} = 2n^2 + 5n + 1$$

```python
def binary_search(arr, key):
    low = 0
    high = len(arr) - 1
    while low <= high:
        mid = low + (high - low) // 2
        if arr[mid] == key:
            return mid
        elif arr[mid] < key:
            low = mid + 1
        else:
            high = mid - 1
    return -1
```
