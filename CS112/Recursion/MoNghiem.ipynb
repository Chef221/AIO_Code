# Phương pháp Đoán nghiệm (Substitution Method)


**Bản chất của phương pháp**

- Thay vì phải đi khai triển, ta sẽ đoán luôn kết quả, sau đó dùng Toán học quy nạp để chứng minh 'cái đoán của mình là đúng'
- Phương pháp này gồm 3 bước chính:

**1. Đoán nghiệm (Guessing)**

Nhìn vào phương trình đệ quy và dùng trực giác (hoặc kinh nghiệm từ các bài toán tương tự) để đoán xem độ phức tạp $T(n)$ sẽ giới hạn bởi hàm $f(n)$ nào
- Các hàm quen thuộc thường là: $\log n, n, n \log n, n^2, 2^n,...$
- Ví dụ: Nếu thấy phương trình bị chia đôi và cộng thêm một hằng số $T(n) = 2T(n/2) + n$, ta có thể đoán độ phức tạp của nó giống với thuật toán Merge Sort là $O(n\log n)$
- Khi dự đoán $T(n) = O(f(n))$, mục tiêu của ta là phải chứng minh được bất đẳng thức: $T(n) \le c \cdot f(n)$ (với $c$ là một hằng số dương nào đó chưa biết)

**2. Bước 2: Giả thiết quy nạp (Inductive Hypothesis)**

Theo nguyên lý quy nạp toán học, nếu muốn chứng minh điều gì đó đúng với n, ta sẽ giả sử nó đã đúng với mọi giá trị $k < n$
- Giả sử: $T(k) \le c \cdot f(k)$ với mọi $k < n$.
- Cụ thể trong các bài toán chia để trị, ta thường giả sử nó đúng với $k = n/2$ hoặc $k = n-1$.

**3. Bước quy nạp (Inductive Step) và Tìm hằng số**

Ta lấy phương trình gốc $T(n)$, thay giả thiết quy nạp từ Bước 2 vào, và biến đổi toán học sao cho kết quả cuối cùng ra được dạng $\le c \cdot f(n)$. Trong quá trình biến đổi này, bạn sẽ tìm ra được điều kiện của hằng số $c$.

# Ví dụ minh họa

Giả sử ta có phương trình: 
$$T(n) = 2T(n/2) + n$$

 (với $T(1) = 1$)

**1. Đoán nghiệm**

Minh đoán bài này có độ phức tạp là $O(n \log n)$
- Nghĩa là cần chứng minh $T(n) \le c \cdot n \log n$ (Với c là hằng số cần tìm)  

**2. Giả thiết quy nạp**

Giả sử dự đoán của minh đúng với $k = n/2$, tức là:
$$T(n/2) \le c \cdot (n/2) \log_2(n/2)$$

Giải thích tại sao lại chọn $k = n/2$
- Phương trình đệ quy gọi cái gì, ta giả sử cái đó
    - Phương trình gốc:
    $$T(n) = 2T(n/2) + n$$
    - Ta có thể thấy, để tính được giá trị của $T(n)$, thuật toán bắt buộc phải biết được giá trị của $T(n/2)$. Nó hoàn toàn không quan tâm đến $T(n-1), T(n-3), T(n-4),...$ ở ngay bước đó
    - Mục tiêu của bước 3 là ta phải triệt tiêu được chữ $T$ ở vế phải để biến nó thành một bất đẳng thức chỉ chứa n. Để làm được điều đó, ta phải tìm cách thế một cái gì đó vào chỗ của $T(n/2)$
    - Do đó ở bước này, ta chủ động chọn $k = n/2$ để tạo ra $T(n/2) \le c \cdot (n/2) \log_2(n/2)$ và nạp nó thẳng vào phương trình.



**3. Bước quy nạp (Chứng minh)**

Ta lấy phương trình gốc và thế giả thiết vào:
$$T(n) = 2 \cdot T(n/2) + n$$
$$T(n) \le 2 \cdot \left[ c \cdot \frac{n}{2} \log_2\left(\frac{n}{2}\right) \right] + n$$
$$T(n) \le c \cdot n \cdot (\log_2 n - \log_2 2) + n$$

(áp dụng tính chất logarit: $\log(a/b) = \log a - \log b$)

$$T(n) \le c \cdot n \cdot (\log_2 n - 1) + n$$
$$T(n) \le c \cdot n \log_2 n - c \cdot n + n$$
$$T(n) \le c \cdot n \log_2 n - n(c - 1)$$

Đến đây, để bất đẳng thức $T(n) \le c \cdot n\log n$ xảy ra, thì cái cụm bị trừ đi là $n(c-1)$ phải lớn hơn hoặc bằng 0


Tại sao lại xuất hiện điều kiện $n(c-1) \ge 0$

- Mục tiêu:
$$T(n) \le c \cdot n \log_2 n$$

Đây chính là cái đích cuối cùng chúng ta bắt buộc phải chứng minh được. Mọi phép biến đổi toán học đều phải hướng về việc tạo ra đúng cái hình dáng $c \cdot n \log_2 n$ này nằm gọn gàng bên vế phải.
- Phân tích kết quả
$$T(n) \le \underbrace{c \cdot n \log_2 n}_{\text{Cái ta muốn}} - \underbrace{n(c-1)}_{\text{Phần thừa ra}}$$

Vế phải hiện tại đang là "Cái ta muốn" TRỪ ĐI "Phần thừa ra".

- Bất đẳng thức

    - Trong toán học, có một tính chất hiển nhiên như sau: Một số $A$ đem trừ đi một số dương (hoặc số 0) thì chắc chắn kết quả sẽ nhỏ hơn hoặc bằng chính số $A$ ban đầu.
    - Công thức toán học: Nếu $X \ge 0$ thì $A - X \le A$.
    - Ta có chuỗi bất đẳng thức sau:
    $$T(n) \le A - X$$
    - Mà để chứng minh thành công, ta cần $T(n) \le A$.
Vậy, để bắt cầu sang được $A$, thì biểu thức $A - X$ bắt buộc phải $\le A$.
Và để $A - X \le A$ xảy ra, thì lượng bị trừ đi ($X$) phải là một số không âm ($X \ge 0$).

- Giải điều kiện để tìm $c$
$$n(c-1) \ge 0$$

- Vì $n$ là kích thước của bài toán (số lượng phần tử mảng, số dòng code...), nên $n$ luôn là một số dương ($n > 0$).
- Do đó, để tích số này $\ge 0$, thì bản thân cái ngoặc $(c-1)$ phải $\ge 0$.
$$\implies c - 1 \ge 0$$
$$\implies c \ge 1$$
