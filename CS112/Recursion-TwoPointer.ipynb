# (1) Recursion

**1.Recursion**

- A programming technique that breaks down a complex problem into smaller manageble pieces
- Recursive solutions solve a problem by applying the same algorithm to each piece then combining the results


```python
# Sum function
def countSum(arr):
    if arr == []:
        return 0
    else:
        smaller_problem = arr[1:]
        smaller_result = countSum(smaller_problem)
        return arr[0] + smaller_result
    
```

```python
# Factorial Function
def countFactorial(n):
    if n == 0:
        return 1
    else:
        smaller_fac = n - 1
        smaller_res = countFactorial(smaller_fac)
        return n * smaller_res
print(countFactorial(4))
```

```python
# Fibonacci Number
def Fibonacci(n):
    if n == 0:
        return 0
    elif n == 1:
        return 1
    else:
        return Fibonacci(n - 1) + Fibonacci(n - 2)
    
print(Fibonacci(4))
```
