```python
import numpy as np

def one_hot(y, c):
    one_hot_z = np.zeros((len(y), c))
    for i in range(len(y)):
        # i: là chỉ số hàng (đại diện cho điểm dữ liệu thứ i)
        # y[i]: là giá trị nhãn thực tế, cũng chính là vị trí cột cần gán giá trị 1
        label_index = y[i]

        # Tại hàng i, gán giá trị 1 cho cột tương ứng với nhãn
        one_hot_z[i, label_index] = 1
    return one_hot_z
```

```python
def one_hot(y, c):
    one_hot_z = np.zeros((len(y), c))
    one_hot_z[np.arange(len(y)), y] = 1
    return one_hot_z
```

```python
y = np.array([0, 2, 1])
c = 3
one_hot_z = one_hot(y, c)
one_hot_z
```

```python
def softmax(X):
    exps = np.exp(X)
    return exps / np.sum(exps)

def stable_softmax(X):
    exps = np.exp(X - np.max(X))
    return exps / np.sum(exps)
```
