# (6) Indexing & Slicing (Truy cập dữ liệu)
- Cách lấy dữ liệu ra khỏi mảng

```python
import numpy as np
'''
Cú pháp: arr[start:stop:step]
    - start: vị trí bắt đầu (lấy cả vị trí này)
    - stop: vị trí kết thúc (không lấy vị trí này)
    - step: mặc định là 1
'''
import numpy as np

arr = np.array([10, 20, 30, 40, 50, 60])
# Index:       0   1   2   3   4   5
# Negative:   -6  -5  -4  -3  -2  -1

# --- INDEXING (Lấy 1 phần tử) ---
print(arr[0])   # 10 (Phần tử đầu tiên)
print(arr[-1])  # 60 (Phần tử cuối cùng)

# --- SLICING (Cắt một đoạn) ---
print(arr[1:4])  # Lấy index 1, 2, 3 -> [20 30 40] (Không lấy index 4)
print(arr[:3])   # Lấy từ đầu đến sát index 3 -> [10 20 30]
print(arr[3:])   # Lấy từ index 3 đến hết -> [40 50 60]

# --- STEP (Bước nhảy) ---
print(arr[::2])  # Lấy từ đầu đến cuối, nhảy 2 bước -> [10 30 50]
print(arr[::-1]) # Đảo ngược mảng -> [60 50 40 30 20 10]
```

```python
arr_2d = np.array([
    [1, 2, 3, 4, 5],       # Hàng 0
    [6, 7, 8, 9, 10],      # Hàng 1
    [11, 12, 13, 14, 15]   # Hàng 2
])

# Lấy một pahnaf tử cụ thể
# Cú pháp: [hàng, cột]
print(arr_2d[1, 2])

# Lấy trọn vẹn một hàng hoặc một cột
# Lấy toàn bộ hàng 1
print(arr_2d[1, :])

# Lấy toàn bộ cột 2
print(arr_2d[:, 2])

# Cắt ma trận con
# Lấy 2 hàng đầu, từ cột 1 đến cột 3
print(arr_2d[:2, 1:4])
```

```python
# Phân biệt [0, 2] và [[0, 2]]
arr = np.array([
    [10, 11, 12],  # Dòng 0
    [20, 21, 22],  # Dòng 1
    [30, 31, 32]   # Dòng 2
])
# Lấy dòng 0 và dòng 2
print(arr[[0, 2]]) 
# -> [[10, 11, 12],
#     [30, 31, 32]]
```

Trường hợp 1: Dùng 1 cặp ngoặc arr[0, 2]
- Đây là Tọa độ (Coordinate).   
    - NumPy hiểu là: "Lấy phần tử ở hàng 0, cột 2".
    - Kết quả: Số vô hướng 12.
Trường hợp 2: Dùng 2 cặp ngoặc lồng nhau arr[[0, 2]]
- Đây là Danh sách các hàng (List of rows).
    - Bạn đang truyền vào một cái List [0, 2] làm chỉ số (index).
    - NumPy hiểu là: "Hãy nhặt cho tôi toàn bộ hàng số 0 và toàn bộ hàng số 2".
    - Kết quả: Một ma trận mới gồm 2 hàng này.

```python
# Clip(): Hàm clip cắt giới hạn các giá trị trong một mảng vào một khoảng 
# giá trị nhất định
a = np.array([1, 5, 8, 10, 3, 6])
result = np.clip(a, a_min=3, a_max=8)
print(result)
```

```python
# Concatenate: Nối các mảng lại với nhau theo một trục nhất định
# axis = 0: Từ trên xuống (Chiều hàng)
# axis = 1: Từ trái qua phải (Chiều cột)
arr_1 = np.array([1, 2, 3])
arr_2 = np.array([4, 5, 6])
arr_concate = np.concatenate((arr_1, arr_2), axis = 0)
print(arr_concate)

arr_1 = np.array([
    [1, 2, 3]
])
arr_2 = np.array([
    [4, 5, 6]
])
arr_concate = np.concatenate((arr_1, arr_2), axis = 0)
print(arr_concate)
print(arr_concate.shape)

arr_concate = np.concatenate((arr_1, arr_2), axis = 1)
print(arr_concate)
print(arr_concate.shape)
```

```python
# Add more axis: Thêm một trục mới vào mảng, mở rộng số chiều của mảng
arr = np.array([1, 2, 3, 4, 5])
print(arr)
print(arr.shape)
arr1 = arr[np.newaxis, :]
arr2 = arr[:, np.newaxis]
print(arr1)
print(arr1.shape)
print(arr2)
print(arr2.shape)
```

```python
# Vectorize(): Chuyển đổi một hàm không phải vector thành một hàm có thể áp
# dụng cho các mảng numpy
def compare(arr1, arr2):
    if arr1 < arr2:
        return 1
    return 0
arr1 = np.array([1, 2, 3, 4, 5])
arr2 = np.array([2, 3, 0, 1, 4])
vectorize_compare = np.vectorize(compare)
vectorize_compare(arr1, arr2)


def calculate_circle_area(radius):
    return np.pi * (radius ** 2) 
radius_arr = np.array([1, 2, 3, 4])
calculate_circle_area_vectorize = np.vectorize(calculate_circle_area)
area_arr = calculate_circle_area_vectorize(radius_arr)
print(area_arr)
```

```python
# Where(): Xác định vị trí của các phần tử trong mảng thỏa mãn một điều kiện gì đó
arr = np.array([0, 1, 2, 3, 4])
out = np.where(arr > 3, arr, 0)
print(out)
```

```python
def matrix_multiplication(matrix1, matrix2):
    matrix1_nrows = len(matrix1)
    matrix1_ncols = len(matrix1[0])
    matrix2_nrows = len(matrix2)
    matrix2_ncols = len(matrix2[0])

    result = [[0] * matrix2_ncols for i in range(matrix1_nrows)]

    for i in range(matrix1_nrows):
        for j in range(matrix2_ncols):
            for k in range(matrix2_nrows):
                result[i][j] += matrix1[i][k] * matrix2[k][j]

    return result
```
