# Lộ trình xây dựng SVM

# Chương 1: Ý tưởng hình học (Ta muốn gì?)

**1. Bài toán: Phân chia lãnh thổ**
- Bạn có hai đội quân:
    - Đội Xanh: Nằm rải rác ở góc dưới bên trái
    - Đội Đỏ: Nằm rải rác ở góc trên bên phải
- Nhiệm vụ: Bạn cần kẻ một đường thẳng gọi là biên giới để chia tách hoàn toàn hai đội này
- Vấn đề là: Có vô số đường thẳng có thể làm việc này. Bạn có thể kẻ một đường sát sạt đội Xanh, hoặc một đường sát sạt đội Đỏ, hoặc một đường xiên xẹo ở giữa

**2. Chọn đường biên giới xịn nhất (Max Margin)**
- Trong SVM, đường biên giới xịn nhất là đường biên giới an toàn nhất
- Hãy tưởng tượng đường biên giới không phải là một sợi chỉ mỏng manh, mà là một con đường cao tốc
    - Lề đường (Margin): Là độ rộng của con đường đó
    - Luật chơi: Con đường này phải được mở rộng ra hết cỡ có thể, cho đến khi mép đường chạm vào những người đứng gần nhất của cả hai đội

**3. Tại sao phải mở rộng lề đường (Maximize Margin) ?**
- Nếu con đường quá hẹp (Lề nhỏ): Khi có một dữ liệu mới, nếu rơi hơi lệch một chút, rất dễ bị rơi nhầm sang loại khác -> Phân loại sai (Lỗi)
- Nếu con đường rất rộng (Lề lớn): Dữ liệu mới có bị nhiễu, rung lắc hay lệch đi một chút nó vẫn nằm an toàn trong vùng lãnh thổ đúng -> Mô hình tổng quát hóa tốt (Generalization)

**4. Các thành phần chính:**
- Siêu phẳng (Hyperplane)
    - Đây là đường phân chia chính thức
    - Phương trình: $w^T x + b = 0$.
    - Bất kỳ điểm nào nằm trên đường này đều là điểm trung lập.
- Hai đường biên
    - Đây là hai mép của con đường cao tốc.
    - Mép phía đội Xanh: $w^T x + b = -1$.
    - Mép phía đội Đỏ: $w^T x + b = +1$.
    - (Tại sao lại là 1 và -1? Thật ra là số nào cũng được, nhưng các nhà toán học chọn số 1 để tính toán cho đẹp và chuẩn hóa).
- Support Vectors:
    - Đây là những điểm dữ liệu chạm vào đường biên.

    - Chúng giống như những cái "cọc" chống đỡ con đường. Nếu bạn xóa hết các điểm ở xa đi, con đường vẫn y nguyên. Nhưng nếu bạn di chuyển các Support Vector này, con đường sẽ bị thay đổi.

    - Tên gọi SVM: Support Vector Machine = Cỗ máy dựa trên các véc-tơ hỗ trợ (chỉ quan tâm đến các điểm ở biên).

**5. Mối liên hệ: Từ Hình học sang Công thức $||w||$**
- Đây là bước nhảy quan trọng nhất mà nhiều người hay bị rối.
    - Chúng ta muốn Lề (Margin) càng to càng tốt.
    - Trong hình học giải tích, khoảng cách giữa 2 đường thẳng song song ($w^T x + b = 1$ và $w^T x + b = -1$) được tính bằng công thức:
    $$Margin = \frac{2}{||w||}$$

    (Trong đó $||w||$ là độ dài của vector pháp tuyến $w$).

- Suy luận logic:
    - Ta muốn Maximize cái Margin ($\frac{2}{||w||}$).
    - Để phân số $\frac{2}{||w||}$ lớn nhất, thì mẫu số $||w||$ phải nhỏ nhất.
    - Về mặt toán học, thiểu hóa $||w||$ cũng giống như tối thiểu hóa $||w||^2$ (để dễ đạo hàm).

$\rightarrow$ Kết luận: Chính vì ý tưởng hình học "muốn con đường rộng nhất", nên ta mới đi đến công thức toán học "tối thiểu hóa độ dài vector trọng số $||w||$".

# Chương 2: Bài toán Gốc (The Primal Problem)

**1. Hàm mục tiêu: "Làm sao để đường rộng nhất?"**

Từ chương 1, ta đã biết độ rộng của lề đường (Margin) là:
$$Margin = \frac{2}{||w||}$$

- Logic: Để Margin Max, thì mẫu số $||w||$ phải Min.
- Kỹ thuật Toán học:
    - Thay vì tìm Min của $||w||$ (chứa căn bậc hai, khó tính đạo hàm), các nhà toán học đổi sang tìm Min của $||w||^2$.
    - Họ thêm số $\frac{1}{2}$ vào trước để khi đạo hàm ($x^2$ đạo hàm ra $2x$), số 2 sẽ triệt tiêu với $\frac{1}{2}$, giúp công thức đẹp hơn.

$\rightarrow$ Kết luận 1 (Hàm mục tiêu):
$$\min \frac{1}{2}||w||^2$$

- Cũ (Hard Margin): Tuyệt đối không chấp nhận lỗi.
    - Mục tiêu: $\min \frac{1}{2}||w||^2$
    - Ràng buộc: $y_i(w^T x_i + b) \geq 1$ (Bắt buộc phải đúng).
- Mới (Soft Margin): Chấp nhận lỗi, nhưng phải trả giá.
    - Chúng ta cho phép mỗi điểm dữ liệu được phép "lấn vạch" một đoạn là $\xi_i$.- Hàm mục tiêu mới:
    $$\min \frac{1}{2}||w||^2 + C \sum_{i=1}^{n} \xi_i$$
    - Ràng buộc mới:
    $$y_i(w^T x_i + b) \geq 1 - \xi_i$$


**2. Ràng buộc: "Luật giao thông trên đường cao tốc"**

Ta không thể cứ thu nhỏ $||w||$ về 0 một cách vô tội vạ được (nếu $w=0$, Margin = vô cực, nhưng chẳng phân loại được gì cả).

Ta phải đảm bảo "Luật giao thông": Không ai được phép đi vào giữa lòng đường (vùng Margin).

- Với đội Xanh (Class +1): Phải nằm ở lề trên hoặc cao hơn.Với đội Xanh (Class +1): Phải nằm ở lề trên hoặc cao hơn.
$$w^T x_i + b \geq 1$$
- Với đội Đỏ (Class -1): Phải nằm ở lề dưới hoặc thấp hơn.
$$w^T x_i + b \leq -1$$

Kỹ thuật gộp công thức (Tuyệt chiêu):
Viết 2 dòng trên thì dài dòng quá. Các nhà toán học gộp nó lại thành 1 dòng duy nhất bằng cách nhân với nhãn $y_i$:
- Nếu $y_i = 1$ (Đội Xanh): Ta giữ nguyên $1 \cdot (w^T x_i + b) \geq 1$. (Đúng)
- Nếu $y_i = -1$ (Đội Đỏ): Ta nhân -1 vào bất đẳng thức $\leq -1$. Khi nhân số âm, dấu bất đẳng thức đổi chiều $\rightarrow$ $(-1) \cdot (w^T x_i + b) \geq 1$. (Cũng đúng!)

$\rightarrow$ Kết luận 2 (Ràng buộc duy nhất):$$y_i(w^T x_i + b) \geq 1$$

(Nghĩa là: Dù anh là Xanh hay Đỏ, nếu nhân vị trí của anh với nhãn của anh, kết quả luôn phải lớn hơn hoặc bằng 1).

**3. Tổng hợp: Bài toán Primal hoàn chỉnh**

Bây giờ ghép phần 1 và phần 2 lại, ta có bài toán SVM kinh điển (dạng Primal) mà bạn thấy trong mọi giáo trình:
- Tìm $w$ và $b$ để:
    - Minimize: $\quad f(w) = \frac{1}{2}||w||^2$  (Để đường rộng nhất)
    - Subject to (Thỏa mãn): $\quad y_i(w^T x_i + b) \geq 1$ với mọi $i = 1..n$ (Để không ai vi phạm luật giao thông)

**4. Tại sao chúng ta không dừng ở đây?**

Đến đây bạn sẽ hỏi: "Có công thức rồi, sao không ném vào máy tính giải luôn mà phải đẻ ra Chương 3 (Lagrange) làm gì?"

Lý do là: 
- Đây là bài toán tối ưu có ràng buộc bất đẳng thức (Quadratic Programming with Inequality Constraints). Giải trực tiếp nó khá "khó chịu" và tốn kém tài nguyên tính toán.
- Quan trọng hơn: Dạng Primal này phụ thuộc vào số chiều của $w$. Nếu dữ liệu có 1 triệu chiều (hoặc vô hạn chiều), máy tính sẽ "chết đứng" vì không thể tính nổi $w$.
- Nó chưa giải quyết được chuyện Dữ liệu phi tuyến (đường cong).

# Chương 3: Phương pháp Lagrange (Lagrange Multipliers)

Trong toán học, giải một bài toán tìm Min/Max mà bị vướng bởi các điều kiện ràng buộc (bất đẳng thức) là cực kỳ khó chịu. Phương pháp Lagrange là một "mẹo" thông minh để biến bài toán có ràng buộc thành bài toán không ràng buộc.

**1. Ý tưởng "Phạt tiền" (Penalty)**

Hãy tưởng tượng bạn đang lái xe (đi tìm điểm cực tiểu $w$).
- Luật cũ (Chương 2): "Cấm vượt đèn đỏ". Nếu bạn vượt, cảnh sát sẽ chặn bạn lại ngay lập tức. (Đây là Ràng buộc cứng).
- Luật mới (Lagrange): "Cứ vượt thoải mái, nhưng mỗi lần vượt sẽ bị phạt tiền cực nặng". (Đây là Ràng buộc mềm).

Hàm Lagrange ($L$) chính là tổng chi phí bạn phải trả:
$$L = \text{Chi phí xăng xe} + \text{Tiền phạt}$$

Trong SVM, nó được viết là:
$$L(w, b, \alpha) = \underbrace{\frac{1}{2}||w||^2}_{\text{Mục tiêu gốc}} - \sum_{i=1}^{n} \alpha_i \underbrace{[y_i(w^T x_i + b) - 1]}_{\text{Phần kiểm tra luật}}$$
- $\alpha_i$ (Alpha): Chính là "mức phạt" (Lagrange Multiplier). $\alpha_i \geq 0$.
- Cụm trong ngoặc vuông $[\dots]$: Là tình trạng tuân thủ luật.
    - Nếu điểm dữ liệu nằm đúng (an toàn): Cụm này $> 0$.
    - Nếu điểm dữ liệu nằm sai (vi phạm): Cụm này $< 0$.

**2. Trò chơi Min-Max (Saddle Point)**

Đây là điểm tinh tế nhất. Hàm Lagrange tạo ra một cuộc đấu trí giữa 2 phe:
- Phe Xây dựng (Chúng ta - muốn Min $L$): Ta muốn chỉnh $w$ và $b$ sao cho tổng chi phí $L$ là thấp nhất.
- Phe Phá hoại (Thiên nhiên - muốn Max $L$): Nó được quyền chỉnh $\alpha$ (mức phạt) để làm cho $L$ to nhất có thể.

Tại sao lại có trò chơi này? Hãy xem điều gì xảy ra nếu ta vi phạm luật:
- Giả sử ta chọn một $w$ tồi khiến một điểm dữ liệu bị vi phạm (cụm trong ngoặc bị âm).
- Phe Phá hoại sẽ thấy ngay sơ hở đó. Nó sẽ tăng $\alpha$ lên dương vô cùng ($+\infty$).
- Vì $(\text{số âm}) \times (-\infty) = +\infty$, nên $L$ sẽ vọt lên vô cùng.
- Kết quả: Để tránh bị phạt vô cùng, Phe Xây dựng (chúng ta) bắt buộc phải chọn $w$ sao cho thỏa mãn mọi ràng buộc ngay từ đầu.

$\rightarrow$ Kết luận: Bài toán $\min_w \max_\alpha L$ tương đương hoàn toàn với bài toán gốc ở Chương 2, nhưng dễ xử lý đạo hàm hơn.

**3. Điều kiện KKT: "Sự thưa thớt" (Sparsity)**

Sau khi thiết lập hàm $L$, làm sao biết điểm nào là cực trị? Ta cần bộ quy tắc KKT (Karush-Kuhn-Tucker).

Trong 5 điều kiện KKT (như slide bạn gửi), có một điều kiện quan trọng nhất gọi là Complementary Slackness (Độ lệch bù). Đây là chìa khóa hiểu bản chất SVM:

$$\alpha_i \times [y_i(w^T x_i + b) - 1] = 0$$

Phương trình này có nghĩa là tích của 2 số phải bằng 0, nên ít nhất một trong hai số phải bằng 0. Điều này chia dữ liệu thành 2 loại:
- Trường hợp 1: Điểm nằm an toàn (Xa bờ)
    - Lúc này: $y_i(w^T x_i + b) > 1$ (vượt xa mức 1).
    - Cụm trong ngoặc khác 0.
    - Bắt buộc $\alpha_i = 0$.
    - Ý nghĩa: Điểm này không quan trọng, không bị phạt, không ảnh hưởng đến mô hình. (Đây là đại đa số dữ liệu).
- Trường hợp 2: Điểm nằm ngay mép đường (Support Vectors)
    - Lúc này: $y_i(w^T x_i + b) = 1$.
    - Cụm trong ngoặc bằng 0.
    - Lúc này $\alpha_i$ có thể lớn hơn 0.
    - Ý nghĩa: Đây là những điểm "trụ cột". Chúng có trọng số dương và quyết định hình dáng của đường phân chia.
    

# Giải thích Alpha ($\alpha$) là gì?

**1. Lý thuyết**

- Tên gọi toán học: $\alpha$ là Nhân tử Lagrange (Lagrange Multiplier).
- Ý nghĩa thực tế: $\alpha_i$ đại diện cho "quyền lực" hoặc "tầm quan trọng" của điểm dữ liệu thứ $i$ đối với đường phân chia.
    - Giá trị $\alpha_i$ càng lớn $\rightarrow$ Điểm dữ liệu đó càng "khó tính", nó đang tác động lực rất mạnh để đẩy đường biên giới ra xa nó.
    - Giá trị $\alpha_i = 0$ $\rightarrow$ Điểm dữ liệu đó "vô hình", không có tiếng nói, không đóng góp gì vào việc xây dựng mô hình.

**2. Tại sao chỉ Support Vector mới có $\alpha > 0$?**

Lý do nằm ở một luật bất di bất dịch trong tối ưu hóa, gọi là Điều kiện bù (Complementary Slackness) trong định lý KKT.

Công thức của nó như sau:
$$\alpha_i \times \underbrace{[y_i(w^T x_i + b) - 1]}_{\text{Khoảng cách an toàn}} = 0$$

Phương trình này nói rằng: Tích của hai số phải bằng 0. Điều này dẫn đến hai trường hợp (và chỉ hai trường hợp) xảy ra:

Trường hợp 1: "Dân thường" (Points lying strictly on the correct side)
- Đây là những điểm nằm sâu trong lãnh thổ của mình, cách rất xa đường biên giới.
- Lúc này: $y_i(w^T x_i + b) > 1$ (Vượt mức an toàn 1).
- $\rightarrow$ Cụm trong ngoặc vuông Khác 0 (Ví dụ: bằng 0.5, 2, 10...).
- Để tích bằng 0, bắt buộc $\alpha_i$ PHẢI bằng 0.
- Kết luận: Những điểm an toàn (chiếm 90% dữ liệu) có $\alpha = 0$. Chúng hoàn toàn bị ngó lơ.

Trường hợp 2: "Lính gác" (Support Vectors)
- Đây là những điểm nằm ngay trên mép đường (lề đường).
- Lúc này: $y_i(w^T x_i + b) = 1$.
- $\rightarrow$ Cụm trong ngoặc vuông Bằng 0.
- Vì cụm ngoặc đã bằng 0 rồi, nên $\alpha_i$ được phép khác 0 (cụ thể là dương).
- Kết luận: Chỉ những điểm nằm trên biên mới có quyền có $\alpha > 0$.

**3. Hình tượng hóa cho dễ nhớ**

Hãy tưởng tượng đường phân chia (Decision Boundary) giống như một tấm ván gỗ được giữ thăng bằng nhờ các lò xo.
- Các điểm dữ liệu ($\alpha = 0$): Là những người đứng xa tấm ván. Họ chỉ đứng nhìn thôi, không chạm vào ván. Vì không chạm, nên lực tác động của họ lên ván bằng 0. (Họ không quan trọng).
- Support Vectors ($\alpha > 0$): Là những người đứng sát sạt tấm ván và dùng tay đẩy tấm ván để giữ nó không bị đổ về phía mình.
    - Lực đẩy của cánh tay họ chính là giá trị Alpha.
    - Người nào càng khỏe, đẩy càng mạnh ($\alpha$ càng lớn), thì tấm ván càng phải nể sợ và tránh xa người đó ra.

# Chương 4: Bài toán Đối ngẫu (The Dual Problem)

Đây là chương được mệnh danh là "Màn ảo thuật" của SVM. Tại sao lại gọi như vậy? Bởi vì chúng ta sẽ làm một động tác biến hình: Biến mất hoàn toàn $w$ và $b$, chỉ giữ lại $\alpha$.

**1. Nguyên liệu đầu vào (Từ Chương 3)**

Chúng ta đang có hàm Lagrange khổng lồ:
$$L(w, b, \alpha) = \frac{1}{2}||w||^2 - \sum_{i=1}^{n} \alpha_i [y_i(w^T x_i + b) - 1]$$

Nhiệm vụ của chúng ta (Phe Xây dựng) là tìm Min của hàm này theo $w$ và $b$.
- Trong giải tích, muốn tìm cực tiểu (Min) của hàm số theo biến nào, ta đạo hàm theo biến đó và cho bằng 0.


**2. Bước 1: Triệt tiêu $w$ và $b$ (Lấy đạo hàm)**

Chúng ta sẽ vặn hai cái núm điều chỉnh là $w$ và $b$ để tìm điểm thấp nhất.
- Đạo hàm theo $w$:
    - Đạo hàm của $\frac{1}{2}||w||^2$ là $w$.
    - Đạo hàm của cụm phía sau theo $w$ là $\sum \alpha_i y_i x_i$.
    - Cho bằng 0:
    $$w - \sum_{i=1}^{n} \alpha_i y_i x_i = 0 \Rightarrow \boxed{w = \sum_{i=1}^{n} \alpha_i y_i x_i}$$

    Ý nghĩa cực hay: Vector trọng số $w$ (vector pháp tuyến của siêu phẳng) thực chất chỉ là tổng hợp tuyến tính của các điểm dữ liệu $x_i$ (nhân với trọng số $\alpha_i y_i$).
    - Những điểm nào có $\alpha_i = 0$ (không phải Support Vector) thì không đóng góp gì vào $w$.
    - Chỉ những Support Vector ($\alpha_i > 0$) mới cấu thành nên $w$.

- Đạo hàm theo $b$:
    - Trong cụm $\sum \alpha_i y_i (w^T x_i + b)$, phần dính tới $b$ là $\sum \alpha_i y_i b$.
    - Đạo hàm theo $b$ sẽ còn lại $\sum \alpha_i y_i$.Cho bằng 0:
    $$\boxed{\sum_{i=1}^{n} \alpha_i y_i = 0}$$

    Ý nghĩa: Tổng các hệ số $\alpha$ của lớp Dương (+1) phải cân bằng với tổng các hệ số $\alpha$ của lớp Âm (-1).

**3. Bước 2: Thế ngược (Substitution - Màn ảo thuật)**

Bây giờ, ta mang 2 cái hộp đóng khung ở trên thế ngược lại vào hàm $L$ ban đầu.
$$L = \frac{1}{2}\underbrace{(w^T w)}_{||w||^2} - \sum \alpha_i y_i \underbrace{(w^T x_i)}_{Thay w vào đây} - \sum \alpha_i y_i b + \sum \alpha_i$$

Khi bạn thay thế $w = \sum \alpha_i y_i x_i$ vào, các số hạng sẽ bắt đầu triệt tiêu nhau
    - Cụm chứa $b$ sẽ biến mất vì $\sum \alpha_i y_i = 0$.
    - Cụm chứa $w$ sẽ rút gọn lại thành một biểu thức chỉ chứa $\alpha$ và tích vô hướng $x^T x$.

**4. Kết quả: Bài toán Dual hoàn chỉnh**

Sau khi rút gọn xong xuôi, ta được một bài toán hoàn toàn mới. Lúc này, ta không còn tìm Min nữa (vì đã tìm rồi), mà chuyển sang nhiệm vụ của Phe Phá hoại: Tìm Max theo $\alpha$.

Bài toán Dual (Chương 4):

- Maximize:
$$Q(\alpha) = \sum_{i=1}^{n} \alpha_i - \frac{1}{2} \sum_{i=1}^{n} \sum_{j=1}^{n} \alpha_i \alpha_j y_i y_j (x_i^T x_j)$$
- Subject to (Ràng buộc):
    - $\sum_{i=1}^{n} \alpha_i y_i = 0$
    - $0 \leq \alpha_i \leq C$ (Nếu là Soft Margin) hoặc $0 \leq \alpha_i$ (Nếu là Hard Margin)

**5. Tại sao Chương 4 này lại "Thần thánh"?**

Dữ liệu $x$ chỉ xuất hiện dưới dạng tích vô hướng $(x_i^T x_j)$.

- Nghĩa là: 
    - Để giải bài toán này, máy tính không cần biết tọa độ cụ thể của $x_i$ hay $x_j$.
    - Máy tính chỉ cần biết độ tương đồng (similarity) giữa chúng, tức là con số kết quả của phép nhân $x_i^T x_j$.

- Điều này dẫn đến Kernel Trick:
Nếu ta muốn nâng $x$ lên không gian 1 tỷ chiều ($\phi(x)$), ta chỉ cần thay cái tích vô hướng $(x_i^T x_j)$ bằng một hàm Kernel $K(x_i, x_j)$.

Ví dụ Kernel RBF (Radial Basis Function):
$$K(x_i, x_j) = \exp(-\gamma ||x_i - x_j||^2)$$

Hàm này tính toán cực nhanh ở không gian 2D, nhưng kết quả của nó lại tương đương với tích vô hướng ở không gian vô hạn chiều!

# Chương 5: Giải và Kết luận

**Bước 1: Khôi phục vector pháp tuyến $w$ (Chỉ áp dụng cho trường hợp Tuyến tính)**

Nếu bạn dùng SVM tuyến tính (Linear SVM - không dùng Kernel), chúng ta có thể tính ra $w$ một cách tường minh.
- Công thức (từ đạo hàm Chương 4):
$$w = \sum_{i=1}^{n} \alpha_i y_i x_i$$
- Ý nghĩa:
    - Vector $w$ là tổng hợp lực của các điểm dữ liệu.
    - Vì hầu hết các $\alpha_i = 0$, nên thực tế bạn chỉ cần cộng các Support Vector lại (nhân với nhãn $y$ và độ lớn $\alpha$ của nó).
    - Ví dụ: Nếu có 3 Support Vector, $w$ sẽ là sự pha trộn của 3 vector đó. Hướng của $w$ sẽ bị kéo về phía những thằng có $\alpha$ lớn nhất.

**Bước 2: Khôi phục hệ số tự do $b$ (Bias)**

Sau khi có $w$, đường thẳng $w^T x + b = 0$ vẫn chưa cố định vị trí (nó có thể trượt lên trượt xuống). Ta cần tìm $b$ để "ghim" nó lại đúng chỗ.
- Chiến thuật: Dựa vào các Support Vector.
Ta biết rằng với bất kỳ Support Vector $x_k$ nào (có $0 < \alpha_k < C$), nó phải nằm đúng trên lề đường. Tức là:
$$y_k(w^T x_k + b) = 1$$
- Tính toán:
Nhân cả 2 vế với $y_k$ (lưu ý $y_k^2 = 1$):
$$w^T x_k + b = y_k$$
$$\Rightarrow \boxed{b = y_k - w^T x_k}$$
- Thực tế (Numerical Stability):
Về lý thuyết, bạn chỉ cần 1 điểm Support Vector là tính ra $b$. Nhưng do sai số máy tính, người ta thường tính $b$ cho tất cả các Support Vector rồi lấy giá trị trung bình để ra con số $b$ chính xác nhất.

**Bước 3: Dự báo (Inference) - Phần quan trọng nhất**

Bây giờ bạn đã có mô hình hoàn chỉnh $(w, b)$. Có một điểm dữ liệu mới toanh $x_{new}$ đi vào, máy tính sẽ làm gì?

Trường hợp A: SVM Tuyến tính (Linear)
- Máy tính chỉ cần thay số vào công thức đường thẳng:
$$f(x_{new}) = \text{sign}(w^T x_{new} + b)$$
- Nếu kết quả $> 0 \rightarrow$ Class +1.
- Nếu kết quả $< 0 \rightarrow$ Class -1.

Trường hợp B: SVM Phi tuyến (Kernel SVM) - Bí thuật "Nâng chiều"
- Đây là chỗ cực kỳ tinh tế.
Nếu bạn dùng Kernel (ví dụ RBF) để nâng chiều dữ liệu lên vô tận, bạn KHÔNG THỂ tính ra vector $w$ ở Bước 1 được (vì nó có vô hạn chiều, máy tính không chứa nổi).
- Vậy làm sao dự báo? Ta thế trực tiếp công thức của $w$ vào hàm dự báo:
$$f(x_{new}) = w^T x_{new} + b = (\sum_{i=1}^{n} \alpha_i y_i x_i)^T x_{new} + b$$
$$= \sum_{i=1}^{n} \alpha_i y_i \underbrace{(x_i^T x_{new})}_{\text{Tích vô hướng}} + b$$

- Lúc này, nhờ Kernel Trick, ta thay thế tích vô hướng bằng hàm Kernel:
$$\boxed{f(x_{new}) = \sum_{i \in SV} \alpha_i y_i K(x_i, x_{new}) + b}$$

- Ý nghĩa tuyệt vời của công thức này:
Để phân loại một điểm mới $x_{new}$:
    - Máy tính sẽ so sánh độ tương đồng (Kernel $K$) của điểm mới đó với từng Support Vector cũ ($x_i$).
    - Nó cộng dồn các điểm số tương đồng lại (có trọng số $\alpha_i y_i$).
        - Nếu $x_{new}$ giống các Support Vector phe Dương nhiều hơn $\rightarrow$ Kết quả ra Dương.
        - Nếu $x_{new}$ giống các Support Vector phe Âm nhiều hơn $\rightarrow$ Kết quả ra Âm.
