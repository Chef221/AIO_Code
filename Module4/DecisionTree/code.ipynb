```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.ensemble import RandomForestRegressor
from sklearn.tree import DecisionTreeRegressor
from sklearn.preprocessing import OrdinalEncoder
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error, mean_squared_error
from sklearn.tree import plot_tree
```

# Load dataset

```python
dataset_path = r'C:\Code_AI\AIO_Code\Module4\DecisionTree\Housing.csv'
df = pd.read_csv(dataset_path)
df.head()
```

```python
df.info()
```

# Deal with categorical variables

```python
categorical_cols = df.select_dtypes(
    include=['object']
).columns.to_list()
categorical_cols
```

```python
for col_name in categorical_cols:
    n_categories = df[col_name].nunique()
    print(f'Number of categories in {col_name}: {n_categories}')
```

```python
ordinal_encoder = OrdinalEncoder()
encoded_categorical_cols = ordinal_encoder.fit_transform(
    df[categorical_cols]
)
```

```python
encoded_categorical_df = pd.DataFrame(
    encoded_categorical_cols,
    columns=categorical_cols
)
encoded_categorical_df.head()
```

```python
numerical_df = df.drop(categorical_cols, axis = 1)
numerical_df.head()
```

```python
encoded_df = pd.concat(
    [numerical_df, encoded_categorical_df], axis = 1
)
encoded_df.head()
```

# Normalization

```python
normalizer = StandardScaler()
dataset_arr = normalizer.fit_transform(encoded_df)
dataset_arr
```

# Split X, y

```python
X = dataset_arr[:,1:]
y = dataset_arr[:,0]
print(f'Indepedent Variables shape: {X.shape}')
print(f'Dependent Variable shape: {y.shape}')
```

```python
X_train, X_val, y_train, y_val = train_test_split(
    X, y,test_size=0.3,random_state=1,shuffle=True
)
```

# Train model   

```python
RD_regressor = RandomForestRegressor(
    random_state=1
)
RD_regressor.fit(X_train, y_train)
```

```python
DT_regressor = DecisionTreeRegressor(
    random_state=1
)
DT_regressor.fit(X_train, y_train)
```

# Evaluation

```python
y_pred = DT_regressor.predict(X_val)

mae = mean_absolute_error(y_val, y_pred)
mse = mean_squared_error(y_val, y_pred)

print(f'Evaluation results on validation set:')
print(f'Mean Absolute Error: {mae}')
print(f'Mean Squared Error: {mse}')
```

```python
y_pred = RD_regressor.predict(X_val)

mae = mean_absolute_error(y_val, y_pred)
mse = mean_squared_error(y_val, y_pred)

print(f'Evaluation results on validation set:')
print(f'Mean Absolute Error: {mae}')
print(f'Mean Squared Error: {mse}')
```

# Visualization

```python
regressor = DecisionTreeRegressor(
    random_state=1,
    max_depth=3
)
regressor.fit(X_train, y_train)
```

```python
fig = plt.figure(figsize=(10,10))
plot_tree(regressor)
```
